# Evaluation of the atomic claim extraction method

Proceedure is as follows:
1. Get LLM answers with citations (using the ALCE framework)
2. Sample ~50 of these answers
3. Get human atomic claims
4. Get automatic atomic claims (as given by this library)
5. Measure alignment

```python
import json
import random

random.seed(42)

data = "/home/wallat/citation_eval/ALCE/result/asqa-Meta-Llama-3-8B-Instruct-gtr-shot1-ndoc3-43.json"

data = json.load(open(data))
print(data['data'][0]['question'])
print(data['data'][0]['docs'])
print(data['data'][0]['output'])

print(len(data['data']))

# sample 50 entries from the data
sampled_indices = random.choices(list(range(len(data['data']))), k=50)
print(sampled_indices)

# store question, documents, and answer (the output field)
sampled_date = []

for i in sampled_indices:
    question = data['data'][i]['question']
    docs = data['data'][i]['docs']
    answer = data['data'][i]['output']
    sampled_date.append({'question': question, 'docs': docs, 'answer': answer})

with open('sampled_data.json', 'w') as f:    
    json.dump(sampled_date, f, indent=4)
```
